# LICA-AI Study — 03 — Linear Mixed-Effects Models (LMM)

*Notebook generated: 2026-10-17 | nbformat v4 | Python 3.10+*

**Study:** From Error Correction to Meaning Construction: Evaluating LICA-AI as a Real-Time Cognitive Scaffold for L2 Academic Writing
**Design:** two-group pre-test/post-test (Control vs Experimental), N = 120 participants,
240 rating records (2 raters x 120 participant-phase units), 4 rubric dimensions:
Ideas, Cohesion, Appropriateness, Language.

Run `pip install pandas numpy scipy statsmodels pingouin matplotlib seaborn openpyxl` first.

```python
import pandas as pd, numpy as np, statsmodels.formula.api as smfwide = pd.read_csv('lica_ai_wide.csv')wide['Group'] = (wide['group'] == 'Experimental').astype(int)wide['Time']  = (wide['phase'] == 'Post').astype(int)DIMS = ['Ideas','Cohesion','Appropriateness','Language','Overall']
```

md

```python
res_rows, models = [], {}for dim in DIMS:    m = smf.mixedlm(f'{dim} ~ Group * Time', wide, groups=wide['participant_id']).fit(reml=True)    models[dim] = m    res_rows.append(dict(dimension=dim,        intercept=round(m.params['Intercept'],2),        group_beta=round(m.params['Group'],2),        time_beta=round(m.params['Time'],2),        interaction=round(m.params['Group:Time'],2),        interaction_p=round(m.pvalues['Group:Time'],4),        CI= tuple(round(x,2) for x in m.conf_int().loc['Group:Time']),        ICC_participant=round(m.cov_re.iloc[0,0] / (m.cov_re.iloc[0,0] + m.scale), 3)))lmm_table = pd.DataFrame(res_rows); lmm_table
```
