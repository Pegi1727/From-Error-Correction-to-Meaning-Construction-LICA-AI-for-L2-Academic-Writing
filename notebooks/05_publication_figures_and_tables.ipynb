```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as snssns.set_theme(style='whitegrid', context='paper', font_scale=1.1)plt.rcParams.update({'font.family':'serif','axes.linewidth':0.8})wide = pd.read_csv('lica_ai_wide.csv')DIMS = ['Ideas','Cohesion','Appropriateness','Language','Overall']
```

```python
fig, axes = plt.subplots(1, 5, figsize=(20, 4), sharey=True)for ax, dim in zip(axes, DIMS):    emm = wide.groupby(['group','phase'])[dim].mean().reset_index()    for g, c in [('Control','#4C72B0'), ('Experimental','#DD8452')]:        s = emm[emm['group']==g]        ax.plot(s['phase'], s[dim], marker='o', color=c, label=g)    ax.set_title(dim); ax.set_xlabel('')axes[0].set_ylabel('Mean score'); axes[0].legend(frameon=False)plt.tight_layout(); plt.savefig('fig2_interaction_means.png', dpi=300, bbox_inches='tight')
```

```python
def apa_table(df, caption, fname):    with open(fname, 'w') as f:        f.write(caption + '\n\n'); f.write(df.to_markdown(index=False))    print(caption); print(df.to_markdown(index=False))apa_table(pd.read_csv('table4_3_lmm_interaction.csv'), 'Table 4.3. Fixed-effects interaction coefficients (Group x Time)', 'table4_3.md')apa_table(pd.read_csv('table4_4_morris_d.csv'), "Table 4.4. Standardized difference-in-gains effect sizes (Morris' d)", 'table4_4.md')
```
